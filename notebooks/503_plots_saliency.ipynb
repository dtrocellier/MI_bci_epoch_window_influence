```python
import matplotlib.pyplot as plt
import numpy as np
import mne
```

```python
saliency_temporal_per_subject_0 = np.load("Results/saliency_temporal_per_subject_0_4.npy")
saliency_spatial_per_subject_0 = np.load("Results/saliency_spatial_per_subject_0_4.npy")

saliency_temporal_per_subject_05 = np.load("Results/saliency_temporal_per_subject_05_45.npy")
saliency_spatial_per_subject_05 = np.load("Results/saliency_spatial_per_subject_05_45.npy")
```

```python
mean_saliency_temporal_0 = saliency_temporal_per_subject_0.mean(axis=0)
std_saliency_temporal_0 = saliency_temporal_per_subject_0.std(axis=0)

mean_saliency_spatial_0 = saliency_spatial_per_subject_0.mean(axis=0)
std_saliency_spatial_0 = saliency_spatial_per_subject_0.std(axis=0)
```

```python
mean_saliency_temporal_05 = saliency_temporal_per_subject_05.mean(axis=0)
std_saliency_temporal_05 = saliency_temporal_per_subject_05.std(axis=0)

mean_saliency_spatial_05 = saliency_spatial_per_subject_05.mean(axis=0)
std_saliency_spatial_05 = saliency_spatial_per_subject_05.std(axis=0)
```

```python

# plot the saliency map in spatial domain


# load the channel location
montage = mne.channels.make_standard_montage("standard_1020")
ch_names = ['Fz', 'FCz', 'Cz', 'CPz', 'Pz', 'C1', 'C3', 'C5', 'C2', 'C4', 'C6', 'F4', 'FC2', 'FC4', 'FC6', 'CP2', 'CP4',
            'CP6', 'P4', 'F3', 'FC1', 'FC3', 'FC5', 'CP1', 'CP3', 'CP5', 'P3']
ch_pos = montage.get_positions()["ch_pos"]
ch_pos = np.array([ch_pos[ch_name][:2] for ch_name in ch_names])

ch_pos = montage.get_positions()["ch_pos"]
ch_pos = np.array([ch_pos[ch_name][:2] for ch_name in ch_names])
```

```python
fig, ax = plt.subplots(2,2, figsize=(20, 11))

plt.rcParams.update({
    'font.size': 30,               # Base font size
    'axes.titlesize': 25,         # Title font size
    'axes.labelsize': 25,         # Axis label font size
    'xtick.labelsize': 16,        # X-tick label font size
    'ytick.labelsize': 16,        # Y-tick label font size
    'legend.fontsize': 16,        # Legend font size (if used)
})
# Plotting the temporal saliency 0-4
ax[0,0].plot(mean_saliency_temporal_0, c="black")
ax[0,0].fill_between(range(len(mean_saliency_temporal_0)), mean_saliency_temporal_0 - std_saliency_temporal_0, mean_saliency_temporal_0 + std_saliency_temporal_0, color='grey', alpha=0.5)
ax[0,0].set_title("Epoching 0-4 seconds, with visual cue EEG in input \n Classification accuracy 86.82% \n \n Temporal Saliency", color="red")

ax[0,0].set_xticks(np.arange(0, 2049, 256), labels = np.arange(0, 4.01, 0.5))

ax[0,0].set_xlabel("Time since stimulus onset [s]")
ax[0,0].set_ylabel("Saliency in %")


#  Plotting the temporal saliency 0.5-4.5
ax[0,1].plot(mean_saliency_temporal_05, c="black")
ax[0,1].fill_between(range(len(mean_saliency_temporal_05)), mean_saliency_temporal_05 - std_saliency_temporal_05, mean_saliency_temporal_05 + std_saliency_temporal_05, color='grey', alpha=0.5)
ax[0,1].set_xticks(np.arange(0, 2049, 256), labels = np.arange(0.5, 4.51, 0.5))
ax[0,1].set_xlabel("Time since stimulus onset [s]")
ax[0,1].set_ylabel("Saliency in %")

ax[0,1].set_title("Epoching 0.5-4.5 seconds, without visual cue EEG in input \n Classification accuracy 76.11%\n \n  Temporal Saliency", color="blue")


# Plotting the spatial saliency 0-4
im , _  = mne.viz.plot_topomap(mean_saliency_spatial_0, ch_pos, names=ch_names, cmap="Reds", axes=ax[1,0], show=False,  vlim=(0 ,16))
cbar = plt.colorbar(im, ax=ax[1,0], shrink=0.8)
cbar.set_label("Saliency in %", rotation=270, labelpad=30)
for text in ax[1,0].texts:
    text.set_fontsize(8.5)  # Set to desired font size

ax[1,0].set_title("Spatial Saliency", color="red")


# Plotting the spatial saliency 0.5-4.5
im , _  = mne.viz.plot_topomap(mean_saliency_spatial_05, ch_pos, names=ch_names, cmap="Reds", axes=ax[1,1] , show=False , vlim=(0 ,16))
cbar = plt.colorbar(im, ax=ax[1,1], shrink=0.8)
cbar.set_label("Saliency in %", rotation=270, labelpad=30)
# Adjust the font size of the channel names
for text in ax[1,1].texts:
    text.set_fontsize(8.5)  # Set to desired font size

ax[1,1].set_title("Spatial Saliency", color="blue")
plt.tight_layout()
plt.savefig('Results/saliency_map.png', bbox_inches = "tight",  dpi=300)

plt.show()
```
