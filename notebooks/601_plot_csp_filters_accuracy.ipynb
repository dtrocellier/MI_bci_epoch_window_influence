```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np
```

```python
within = pd.read_csv("Results/csp_acc_4s_within.csv", index_col=0)
cross = pd.read_csv("Results/csp_acc_cross_subject.csv", index_col=0)
```

```python
df_csp = pd.read_csv('Results/accuracy_per_subject_offline.csv', index_col=0)
df_csp.sort_values(by='subject', inplace=True)
# df_csp["subject"] = np.arange(len(df_csp['subject']))
# df["accuracy_CSP_LDA"] = df["accuracy"]
```

```python
within["csp"]=df_csp['accuracy']
cross["csp"]= df_csp["accuracy"]
```

```python
within.sort_values(by='csp', ascending=True, inplace=True)
```

```python
within
```

```python

within.sort_values(by='csp', ascending=True, inplace=True)
within.reset_index(drop=True, inplace=True)

# Example code assuming df is your DataFrame and contains the required columns
plt.figure(figsize=(15, 8))


# Deep ConvNet 0.5-4.5s
plt.plot(within.index, within['acc_05_45'], 'bs--',  label='CSP+LDA epoched 0.5 - 4.5s, acc : {:.1f} %'.format(within['acc_05_45'].mean() * 100))

# Deep ConvNet 0-4s
plt.plot(within.index, within['acc_0_4'], 'rs--',  label='CSP+LDA epoched 0 - 4s, acc : {:.1f} %'.format(within['acc_0_4'].mean() * 100))

# # Optional: CSP+LDA (if you have this data)
# if 'accuracy_CSP_LDA' in df.columns:
#     plt.plot(df.index, df['accuracy_CSP_LDA'], 'y^-', linewidth=2, label='CSP+LDA, acc : {:.1f} %'.format(df['accuracy_CSP_LDA'].mean() * 100))

# Labels and formatting
plt.xlabel('Subjects', fontsize=18)
plt.yticks(np.arange(0.4,1.01,0.1), np.arange(40,101,10))
plt.ylabel('Accuracy (%)',  fontsize=18)
plt.title('Model Accuracy per Subject for CSP+LDA trained in a within subject approach',  fontsize=18)
plt.legend(  fontsize=12)
plt.ylim(0.4, 1)
plt.grid(True)

plt.savefig("Results/model_acc_csp_within.png", dpi=300)
plt.show()
```

```python

cross.sort_values(by='csp', ascending=True, inplace=True)
cross.reset_index(drop=True, inplace=True)

# Example code assuming df is your DataFrame and contains the required columns
plt.figure(figsize=(15, 8))


# Deep ConvNet 0.5-4.5s
plt.plot(cross.index, cross['acc_05_45'], 'bs--',  label='CSP+LDA epoched 0.5 - 4.5s, acc : {:.1f} %'.format(cross['acc_05_45'].mean() * 100))

# Deep ConvNet 0-4s
plt.plot(cross.index, cross['acc_0_4'], 'rs--',  label='CSP+LDA epoched 0 - 4s, acc : {:.1f} %'.format(cross['acc_0_4'].mean() * 100))

# # Optional: CSP+LDA (if you have this data)
# if 'accuracy_CSP_LDA' in df.columns:
#     plt.plot(df.index, df['accuracy_CSP_LDA'], 'y^-', linewidth=2, label='CSP+LDA, acc : {:.1f} %'.format(df['accuracy_CSP_LDA'].mean() * 100))

# Labels and formatting
plt.xlabel('Subjects', fontsize=18)
plt.yticks(np.arange(0.4,1.01,0.1), np.arange(40,101,10))
plt.ylabel('Accuracy (%)',  fontsize=18)
plt.title('Model Accuracy per Subject for CSP+LDA trained in a cross subject approach',  fontsize=18)
plt.legend(  fontsize=12)
plt.ylim(0.4, 1)
plt.grid(True)

plt.savefig("Results/model_acc_csp_cross.png", dpi=300)
plt.show()
```
